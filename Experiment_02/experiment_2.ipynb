```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Create Zomato Food Ordering Dataset

data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009],

    'Customer_Name': [
        'Amit', 'Rahul', 'Priya', 'Sneha', 'Amit',
        'Vikram', 'Neha', 'Rohan', 'Kavya'
    ],
    'Age': [
        22, 25, np.nan, 24, 22,
        28, 26, np.nan, 23
    ],
    'Gender': [
        'Male', 'Male', 'Female', 'Female', 'Male',
        'Male', 'Female', 'Male', 'Female'
    ],
    'Restaurant': [
        'Food Hub', 'Spice Villa', 'Pizza Point', 'Food Hub',
        'Food Hub', 'Biryani House', 'Pizza Point',
        'Spice Villa', 'Biryani House'
    ],
    'Cuisine': [
        'Indian', 'Chinese', 'Italian', 'Indian', 'Indian',
        'Indian', 'Italian', 'Chinese', 'Indian'
    ],
    'City': [
        'Mumbai', 'Pune', 'Mumbai', 'Pune', 'Mumbai',
        'Nashik', 'Mumbai', 'Pune', 'Nashik'
    ],
    'Payment_Method': [
        'UPI', 'Card', 'UPI', np.nan, 'UPI',
        'Cash', 'Card', 'UPI', np.nan
    ],
    'Order_Status': [
        'Delivered', 'Delivered', 'Delivered', 'Cancelled',
        'Delivered', 'Delivered', 'Pending', 'Delivered',
        'Delivered'
    ],

    'Quantity': [2, 3, 1, 2, 2, 4, 2, 3, 2],

    'Order_Amount': [
        450, 720, 350, 500, 450,
        900, 600, 750, 480
    ],

    'Rating': [
        4.5, 4.0, 4.2, 3.5, 4.5,
        4.7, np.nan, 4.1, 4.6
    ]
}

df = pd.DataFrame(data)

print("Original Zomato Dataset:")
display(df)
```

```python
csv_file = "zomato_food_ordering_dataset.csv"
df.to_csv(csv_file, index=False)

from IPython.display import FileLink, display
display(FileLink(csv_file))
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Add a duplicate order for preprocessing demonstration

df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

```

```python
print("Dataset after adding duplicate:")
display(df)

print("\nNumber of duplicate records:")
print(df.duplicated().sum())
```

```python
print("Missing Values Before Preprocessing:")
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Rating'] = df['Rating'].fillna(df['Rating'].mean())
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna(
    df['Payment_Method'].mode()[0]
)
```

```python
print("Missing Values After Handling:")
print(df.isnull().sum())
```

```python
print("Duplicate Records Before Removal:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDuplicate Records After Removal:")
print(df.duplicated().sum())

print("\nDataset Shape After Removing Duplicates:")
print(df.shape)
```

```python
label_encoder = LabelEncoder()

categorical_columns = [
    'Gender',
    'Payment_Method',
    'Order_Status'
]

for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])

print("Dataset After Encoding:")
display(df)
```

```python
scaler = StandardScaler()

numerical_columns = [
    'Age',
    'Quantity',
    'Order_Amount',
    'Rating'
]

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)

print("Dataset After Feature Scaling:")
display(df)
```

```python
print("========== FINAL DATASET ==========")
display(df)

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())
```
